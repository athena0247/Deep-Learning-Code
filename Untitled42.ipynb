```python
# LAB 1 BFS
from collections import deque

def bfs(graph, start, goal):
    queue = deque([([start], 0)])
    explored = set()

    while queue:
        path, cost = queue.popleft()
        state = path[-1]

        if state in explored:
            continue

        print(f"\nExploring Node: {state}")
        print(f"Current Path: {path}")

        explored.add(state)

        for child in graph[state]:
            if child not in explored:
                new_path = path + [child]
                print(f" Generated Child Node: {child}")

                if child == goal:
                    print(f"Goal State {goal} found!")
                    return new_path

                queue.append((new_path, cost + 1))

        print("Frontier:", [p[-1] for p, c in queue])
        print("Explored:", explored)

    return None


graph = {
    "A":["B","C"],
    "B":["A","D","E"],
    "C":["A","F","G"],
    "D":["B"],
    "E":["B","H"],
    "F":["C"],
    "G":["C","I"],
    "H":["E"],
    "I":["G"]
}

solution = bfs(graph, "A", "I")

if solution:
    print("\nSolution path:")
    print("->".join(solution))
    print("Path cost:", len(solution) - 1)
else:
    print("\n Failure: No path exits.")
```

```python
# LAB2 DFS
def dfs(graph, start, goal):
    stack = [([start])]
    explored = set()

    while stack:
        path = stack.pop()
        node = path[-1]

        if node in explored:
            continue

        print("\n expanding node:", node)
        print("current path:", path)

        if node == goal:
            print("\n goal state", goal, "found!")
            return path

        explored.add(node)

        for child in graph[node]:
            if child not in explored:
                new_path = path + [child]
                print("generated child:", child)
                stack.append(new_path)

        print("frontier:", [p[-1] for p in stack])
        print("explored:", explored)

    return None


graph = {
    'A':['B','C'],
    'B':['A','D'],
    'C':['A','E','F'],
    'D':['B'],
    'E':['C'],
    'F':['C','G','H'],
    'G':['F'],
    'H':['F','I'],
    'I':['H']
}

solution = dfs(graph, 'A', 'I')

if solution:
    print("\n solution path:")
    print("-->".join(solution))
    print("path cost:", len(solution)-1)
else:
    print("\n failure: no path exists")
```

```python
import heapq

graph = {
    'A': [('B', 1), ('C', 4)],
    'B': [('A', 1), ('D', 5), ('E', 2)],
    'C': [('A', 4), ('F', 3), ('G', 2)],
    'D': [('B', 5)], 'E': [('B', 2), ('H', 3)],
    'F': [('C', 3)], 'G': [('C', 2), ('I', 1)],
    'H': [('E', 3)], 'I': [('G', 1)]
}

h = {'A': 5, 'B': 4, 'C': 3, 'D': 6, 'E': 3,
     'F': 4, 'G': 1, 'H': 2, 'I': 0}

start, goal = 'A', 'I'
frontier = [(h[start], 0, start, [start])]
explored = set()

while frontier:
    f, g, node, path = heapq.heappop(frontier)

    if node in explored:
        continue

    print("\nExpanding Node:", node)
    print("Current Path:", path)
    print(f"g={g}, h={h[node]}, f={f}")

    if node == goal:
        print(f"\nGoal state {goal} found!")
        print("Solution Path:", "->".join(path))
        print("Path Cost:", g)
        break

    explored.add(node)

    for child, weight in graph.get(node, []):
        if child not in explored:
            new_g = g + weight
            new_f = new_g + h[child]
            heapq.heappush(frontier, (new_f, new_g, child, path + [child]))
            print(f"Generated Child :{child} (g={new_g}, h={h[child]}, f={new_f})")

    print("Frontier:", [(n, f) for f, g, n, p in frontier])
    print("Explored:", explored)
    print()
```

```python
P_B = 0.001
P_E = 0.002

P_A = {(True, True): 0.95, (True, False): 0.94,
       (False, True): 0.29, (False, False): 0.001}

P_J = {True: 0.90, False: 0.05}
P_M = {True: 0.70, False: 0.01}

T, F = True, False

# Enumeration for Burglary = True and False
result = {}

for b in (T, F):
    total = 0

    for e in (T, F):
        for a in (T, F):

            pb = P_B if b else 1 - P_B
            pe = P_E if e else 1 - P_E

            pa = P_A[(b, e)] if a else 1 - P_A[(b, e)]

            pj = P_J[a]
            pm = P_M[a]

            total += pb * pe * pa * pj * pm

    result[b] = total

# Normalize
Z = result[T] + result[F]

print("P(Burglary | John, Mary) =", result[T] / Z)
print("P(Not Burglary | John, Mary) =", result[F] / Z)
```

```python
import numpy as np
import matplotlib.pyplot as plt

# --------------------------------------------------
# 1. Define HMM parameters
# --------------------------------------------------
T = np.array([[0.7, 0.3],
              [0.3, 0.7]])

O = {
    1: np.diag([0.9, 0.2]),  # Umbrella = Yes
    0: np.diag([0.1, 0.8])   # Umbrella = No
}

N, d = 20, 3
rng = np.random.default_rng(12)

# --------------------------------------------------
# 2. Generate random hidden states and observations
# --------------------------------------------------
hidden = []
observations = []
state = rng.choice(2, p=[0.5, 0.5])

for i in range(N):
    state = rng.choice(2, p=T[state])
    umbrella = int(rng.random() < [0.9, 0.2][state])
    hidden.append(state)
    observations.append(umbrella)

print("Observations:", observations)

# --------------------------------------------------
# 3. Filtering
# --------------------------------------------------
def filter_step(f, e):
    f = O[e] @ T.T @ f
    return f / f.sum()

f = np.array([0.5, 0.5])
filtered = []

for obs in observations:
    f = filter_step(f, obs)
    filtered.append(f.copy())

filtered = np.array(filtered)

# --------------------------------------------------
# 4. Direct backward recursion for verification
# --------------------------------------------------
def backward(k, t):
    b = np.ones(2)
    for i in range(t - 1, k - 1, -1):
        b = T @ O[observations[i]] @ b
    return b

# --------------------------------------------------
# 5. Incremental fixed-lag smoothing
# --------------------------------------------------
f = filtered[0].copy()
k = 1
t = k + d

B = np.eye(2)
for i in range(k, t):
    B = B @ T @ O[observations[i]]

smoothed = np.full((N, 2), np.nan)
errors = []

print("\nCurrent Day | Smoothed Day | Rainy | Not Rainy | Error")
print("-" * 60)

while True:
    # Compute smoothed distribution
    b = B @ np.ones(2)
    smooth = f * b
    smooth = smooth / smooth.sum()

    smoothed[k - 1] = smooth

    # Verify using direct backward recursion
    b_direct = backward(k, t)
    direct = f * b_direct
    direct = direct / direct.sum()

    error = np.max(np.abs(smooth - direct))
    errors.append(error)

    print(f"{t:11} | {k:12} | {smooth[0]:.4f} "
          f"| {smooth[1]:.4f} | {error:.2e}")

    if t == N:
        break

    # Equation (14.16):
    # B_new = O_old^(-1) T^(-1) B_old T O_new
    B = np.linalg.solve(O[observations[k]],np.linalg.solve(T, B)) @ T @ O[observations[t]]

    # Advance the forward message
    f = filter_step(f, observations[k])
    k += 1
    t += 1

print("\nMaximum verification error:", max(errors))

# --------------------------------------------------
# 6. Plot filtering and smoothing results
# --------------------------------------------------
days = np.arange(1, N + 1)
valid = ~np.isnan(smoothed[:, 0])

fig, ax = plt.subplots(3, 1, figsize=(11, 10),
                       sharex=True)

# Plot 1: Actual hidden states and observations
actual_rain = 1 - np.array(hidden)

ax[0].step(days, actual_rain, where="mid",
           label="Actual Rain State", linewidth=2)
ax[0].scatter(days, observations, marker="x",
              color="red", s=60, label="Umbrella Observation")
ax[0].set_ylabel("State / Observation")
ax[0].set_title("Hidden States and Observations")
ax[0].set_yticks([0, 1])
ax[0].legend()
ax[0].grid(True, alpha=0.3)

# Plot 2: Filtering and fixed-lag smoothing
ax[1].plot(days, filtered[:, 0], "o-",
           label="Filtering P(Rainy)", alpha=0.8)

ax[1].plot(days[valid], smoothed[valid, 0], "s-",
           label=f"Fixed-Lag Smoothing (d={d})",
           linewidth=2)

ax[1].set_ylabel("Probability")
ax[1].set_title("Filtering vs Fixed-Lag Smoothing")
ax[1].set_ylim(0, 1)
ax[1].legend()
ax[1].grid(True, alpha=0.3)

# Plot 3: Verification error
ax[2].semilogy(days[valid],
               np.maximum(errors, 1e-17), "o-")
ax[2].set_xlabel("Day")
ax[2].set_ylabel("Absolute Error")
ax[2].set_title("Incremental vs Direct Backward Smoothing")
ax[2].grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig("HMM_Fixed_Lag_Smoothing.png", dpi=300)
plt.show()
```
